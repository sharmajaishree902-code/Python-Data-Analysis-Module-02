```python
#Handle Missing Values
#First: Create a Dataset with Missing Values
import pandas as pd
import numpy as np

data = {
    "Name": ["Jaishree", "Priya", "Riya", "Anjali", "Kavya"],
    "Age": [19, 20, np.nan, 19, 22],
    "City": ["Jaipur", "Delhi", "Mumbai", np.nan, "Pune"],
    "Salary": [15000, np.nan, 18000, 22000, 25000]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Find Missing Values
print(df.isnull())
```

```python
#Count Missing Values
print(df.isnull().sum())
```

```python
#Remove Missing Values
cleaned_data = df.dropna()

print(cleaned_data)
```

```python
#Replace Missing Values
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df["City"] = df["City"].fillna("Unknown")

print(df)
```

```python
#Check Missing Values Again
print(df.isnull().sum())
```

```python
#Mini Projects for Module 02
#Project1:Student Performance Analysis
import pandas as pd

data = {
    "Name": ["Jaishree", "Priya", "Riya", "Anjali", "Kavya"],
    "Python": [85, 90, 78, 92, 88],
    "Java": [80, 85, 90, 88, 95],
    "SQL": [90, 88, 85, 95, 80]
}

df = pd.DataFrame(data)

df["Total"] = df[["Python", "Java", "SQL"]].sum(axis=1)

df["Average"] = df["Total"] / 3

print("Student Performance Analysis")

print(df)

print("\nTop Performing Student:")

print(df.loc[df["Average"].idxmax()])
```

```python
#Project2:Employee Salary Analysis
import pandas as pd

data = {
    "Employee": ["Aman", "Riya", "Priya", "Rahul", "Anjali"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [25000, 30000, 35000, 40000, 28000]
}

df = pd.DataFrame(data)

print("Employee Dataset")

print(df)

print("\nAverage Salary:")

print(df["Salary"].mean())

print("\nHighest Salary:")

print(df["Salary"].max())

print("\nLowest Salary:")

print(df["Salary"].min())

print("\nDepartment-wise Average Salary:")

print(df.groupby("Department")["Salary"].mean())
```

```python
#Project 3: Sales Data Analysis
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile"],
    "Quantity": [2, 5, 3, 1, 4],
    "Price": [50000, 15000, 20000, 50000, 15000]
}

df = pd.DataFrame(data)

df["Total_Sales"] = df["Quantity"] * df["Price"]

print("Sales Dataset")

print(df)

print("\nTotal Revenue:")

print(df["Total_Sales"].sum())

print("\nBest Selling Product by Revenue:")

print(df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False))
```
